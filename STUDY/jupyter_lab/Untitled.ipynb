```python
from pandas import read_csv
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from random import randint

data = read_csv("kikidon.csv")

# Разделение данных
# Данные для признаков или данные для обучения
x = data[['area', 'bedrooms','age']]
# Данные для цели или целевая переменная для прогнозирования
y = data[['price']]

# Создание выборок
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print("Размеры тренировочной и тестовых выборок")
print(x_train.shape)
print(x_test.shape)

# Создание модели на основе Линейная регрессия
MODEL = LinearRegression()

# Полноценное обучение модели
MODEL.fit(x_train, y_train)

# Предсказание модели...
y_predict = MODEL.predict(x_test)
print("Предсказание модели\n", y_predict)
```
